```python
def convert_to_datetime_index(df: pd.DataFrame, year: int = 2040) -> pd.DataFrame:
    """
    Convert year_part (e.g., 'jul09') and day_part (e.g., '06h00-07h00-utc') to datetime index.
    
    Parameters:
    -----------
    df : pd.DataFrame
        DataFrame with 'year_part' and 'day_part' columns
    year : int
        The year to use for the datetime index
    
    Returns:
    --------
    pd.DataFrame with datetime index
    """
    df = df.copy()
    
    # Parse year_part: 'jul09' -> 'Jul 09' -> datetime
    # Convert to proper date string format
    year_part_dates = pd.to_datetime(
        df['year_part'].str.capitalize() + f' {year}',
        format='%b%d %Y'
    )
    
    # Parse day_part: '06h00-07h00-utc' -> extract start hour (06:00)
    # Extract the first time component before the '-'
    time_str = df['day_part'].str.split('-').str[0]  # Gets '06h00'
    hours = time_str.str.extract(r'(\d+)h')[0].astype(int)
    minutes = time_str.str.extract(r'h(\d+)')[0].astype(int)
    
    # Combine date and time
    df.index = year_part_dates + pd.to_timedelta(hours, unit='h') + pd.to_timedelta(minutes, unit='m')
    
    return df
```
