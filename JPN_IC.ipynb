```python
from google.cloud import bigquery
import pandas as pd
import numpy as np
import plotly.express as px
from itables import init_notebook_mode, show

client = bigquery.Client()
```

```python
query = """
    SELECT *
    FROM `tz-data-dev.japan_occto_bronze.interconnectors`
"""

# Run the query
query_job = client.query(query)

# Convert the result to a pandas DataFrame
df = query_job.to_dataframe()
```

```python
# Convert datetime and resample to hourly
df['datetime_utc'] = pd.to_datetime(df['datetime_utc'])
df_hourly = df.set_index('datetime_utc').groupby('sortedZoneKeys').resample('H')['netFlow'].mean().reset_index()

# Filter to specific date range
start_date = '2024-05-31 15:00:00+00:00'
end_date = '2025-05-31 14:00:00+00:00'
df_hourly = df_hourly[(df_hourly['datetime_utc'] >= start_date) & (df_hourly['datetime_utc'] <= end_date)]

df_hourly.to_csv('JPN_IC_hourly.csv', index=False)
df_hourly
```
